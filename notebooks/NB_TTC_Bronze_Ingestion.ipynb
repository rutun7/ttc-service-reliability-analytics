```python
import requests
import time

# Toronto Open Data CKAN package ID for TTC Subway Delay Data
package_id = "996cfe8d-fb35-40ce-b569-698d51fc683b"

api_url = (
    "https://ckan0.cf.opendata.inter.prod-toronto.ca/"
    f"api/3/action/package_show?id={package_id}"
)

package_retrieved = False

for attempt in range(3):
    try:
        response = requests.get(api_url, timeout=120)
        response.raise_for_status()

        package = response.json()["result"]

        print("Subway package retrieved successfully\n")

        # Display available resources
        for resource in package["resources"]:
            print("Name:", resource.get("name"))
            print("Format:", resource.get("format"))
            print("URL:", resource.get("url"))
            print("-" * 80)

        package_retrieved = True
        break

    except requests.exceptions.RequestException as e:
        print(f"Attempt {attempt + 1} failed:", e)

        if attempt < 2:
            print("Retrying in 10 seconds...")
            time.sleep(10)

if not package_retrieved:
    raise Exception(
        "Subway package retrieval failed after 3 attempts. "
        "Toronto Open Data API is currently unavailable."
    )
```

```python
# Find the current CSV resource for TTC Subway Delay Data since 2025

subway_resource = next(
    resource
    for resource in package["resources"]
    if resource.get("name") == "TTC Subway Delay Data since 2025.csv"
    and resource.get("format", "").upper() == "CSV"
)

subway_url = subway_resource["url"]

print("Selected resource:")
print(subway_resource["name"])
print(subway_url)
```

```python
import requests
import time

# Bronze locations
bronze_folder = "Files/bronze/subway"
bronze_file = "Files/bronze/subway/ttc_subway_delays_since_2025.csv"
temp_file = "/tmp/ttc_subway_delays_since_2025.csv"

# Create Bronze folder
notebookutils.fs.mkdirs(bronze_folder)

download_successful = False

# Download with retry logic
for attempt in range(5):
    try:
        print(f"Download attempt {attempt + 1}...")

        response = requests.get(subway_url, timeout=180)
        response.raise_for_status()

        # Save temporarily
        with open(temp_file, "wb") as file:
            file.write(response.content)

        # Copy into OneLake Bronze layer
        notebookutils.fs.cp(
            f"file:{temp_file}",
            bronze_file,
            True
        )

        print("Subway Bronze ingestion successful")
        print("Saved to:", bronze_file)
        print("File size:", len(response.content), "bytes")

        download_successful = True
        break

    except requests.exceptions.RequestException as e:
        print(f"Attempt {attempt + 1} failed:", e)

        if attempt < 4:
            print("Waiting 15 seconds before retry...")
            time.sleep(15)

if not download_successful:
    raise Exception(
        "Subway download failed after 5 attempts. "
        "Toronto Open Data is currently not serving the resource successfully."
    )
```
